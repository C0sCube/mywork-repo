```python
from google.cloud import documentai

client = documentai.DocumentProcessorServiceClient()

name = "projects/170460173493/locations/asia-south1/processors/7e592cb5b3b0e132"

with open("test.pdf", "rb") as f:
    document = {"content": f.read(), "mime_type": "application/pdf"}

request = {"name": name, "raw_document": document}
result = client.process_document(request=request)

doc = result.document

# FULL TEXT
full_text = doc.text

print("Full Text:\n", full_text)


# ✅ TOKEN LEVEL (WORD + COORDINATES)
for page in doc.pages:
    for token in page.tokens:
        # Extract text using text anchor
        text_anchor = token.layout.text_anchor
        start_index = text_anchor.text_segments[0].start_index
        end_index = text_anchor.text_segments[0].end_index
        word = doc.text[int(start_index):int(end_index)]

        # Bounding box
        vertices = token.layout.bounding_poly.vertices

        print("Word:", word)
        print("Coords:", [(v.x, v.y) for v in vertices])
        print("------")
```
